### Intermediate Level Python  - Visualization

### Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
year = [1950, 1960, 1970, 1980, 1990]
pop = [2.5, 3.0, 3.7, 4.4, 5.3]

print(f"Last Items in year and population are:\n",year[-1])
print(pop[-1])
```

```python
x=np.linspace(0,2*np.pi,100)
y=np.sin(x)
fig,ax=plt.subplots()
ax.plot(x,y)
plt.show()
```

## Labels

```python
plt.plot(year, pop, marker='o')
plt.xlabel('Year')
plt.ylabel('Population (billions)')
plt.title('World Population Over Time')
plt.yticks(np.arange(0, 6, 1))#yticks - want specific axis intervals or readable labels.
plt.show()
```

### Bubble scatter plot

```python
## plt.xticks(ticks, labels)
# alpha controls transparency.
# c means colour.

    # X-axis = GDP per capita
    # Y-axis = Life expectancy
    # Bubble size = Population
    # Bubble colour = col
    # Transparency = alpha

gdp_cap = [974.58, 5937.03, 6223.37, 4797.35, 10088.0]
life_exp = [43.828, 76.423, 72.961, 75.32, 81.235]
col = ['red', 'blue', 'green', 'orange', 'purple']
# Specify c and alpha inside plt.scatter()
plt.scatter(x = gdp_cap, y = life_exp, s = np.array(pop) * 20,c=col,alpha=0.8)

# Previous customizations
plt.xscale('log') 
plt.xlabel('GDP per Capita [in USD]')
plt.ylabel('Life Expectancy [in years]')
plt.title('World Development in 2007')
plt.xticks([1000,10000,100000], ['1k','10k','100k'])
plt.grid(True, which='major', linestyle='-', linewidth='0.5', color='gray')

# Show the plot
plt.show()
```
