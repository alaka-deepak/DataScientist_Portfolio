### Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
year = [1950, 1960, 1970, 1980, 1990]
pop = [2.5, 3.0, 3.7, 4.4, 5.3]

print(f"Last Items in year and population are:\n",year[-1])
print(pop[-1])
```

```python
x=np.linspace(0,2*np.pi,100)
y=np.sin(x)
fig,ax=plt.subplots()
ax.plot(x,y)
plt.show()
```

## Labels

```python
plt.plot(year, pop, marker='o')
plt.xlabel('Year')
plt.ylabel('Population (billions)')
plt.title('World Population Over Time')
plt.yticks(np.arange(0, 6, 1))#yticks - want specific axis intervals or readable labels.
plt.show()
```

### Bubble scatter plot

```python
## plt.xticks(ticks, labels)
# alpha controls transparency.
# c means colour.

    # X-axis = GDP per capita
    # Y-axis = Life expectancy
    # Bubble size = Population
    # Bubble colour = col
    # Transparency = alpha

gdp_cap = [974.58, 5937.03, 6223.37, 4797.35, 10088.0]
life_exp = [43.828, 76.423, 72.961, 75.32, 81.235]
col = ['red', 'blue', 'green', 'orange', 'purple']
# Specify c and alpha inside plt.scatter()
plt.scatter(x = gdp_cap, y = life_exp, s = np.array(pop) * 20,c=col,alpha=0.8)

# Previous customizations
plt.xscale('log') 
plt.xlabel('GDP per Capita [in USD]')
plt.ylabel('Life Expectancy [in years]')
plt.title('World Development in 2007')
plt.xticks([1000,10000,100000], ['1k','10k','100k'])
plt.grid(True, which='major', linestyle='-', linewidth='0.5', color='gray')

# Show the plot
plt.show()
```

```python
import pandas as pd
from io import StringIO

data = """date,type,year,avg_price,size,nb_sold
2015-12-27,conventional,2015,0.95,small,9626901.09
2015-12-20,conventional,2015,0.98,small,8710021.76
2015-12-13,conventional,2015,0.93,small,9855053.66
2015-12-06,conventional,2015,0.89,small,9405464.36
2015-11-29,conventional,2015,0.99,small,8094803.56
2015-11-22,conventional,2015,0.96,small,8900000.25
2015-11-15,conventional,2015,1.02,small,7800000.50
2015-11-08,conventional,2015,0.91,small,9200000.75
2015-12-27,organic,2015,1.45,small,1200000.50
2015-12-20,organic,2015,1.50,small,1150000.25
2015-12-13,organic,2015,1.48,large,850000.75
2015-12-06,organic,2015,1.52,large,900000.50
2016-01-03,conventional,2016,0.97,large,7500000.25
2016-01-10,conventional,2016,1.01,large,7200000.50
2016-01-17,organic,2016,1.55,large,950000.25
2016-01-24,organic,2016,1.49,small,1250000.75
"""

avocado = pd.read_csv(StringIO(data))
avocado["date"] = pd.to_datetime(avocado["date"])

print(avocado.head())
print(avocado.info())
```

```python
# Import matplotlib.pyplot with alias plt
import matplotlib.pyplot as plt

# Look at the first few rows of data
print(avocado.head())

# Get the total number of avocado sold of each size
nb_sold_by_size = avocado.groupby("size")["nb_sold"].sum()

# Create a bar plot of the number of avocado sold by size
nb_sold_by_size.plot(kind="bar")
# Show the plot
plt.show()
```

```python
# Import matplotlib.pyplot with alias plt
import matplotlib.pyplot as plt

# Get the total number of avocado sold on each date
nb_sold_by_date = avocado.groupby("date")["nb_sold"].sum()

# Create a line plot of the number of avocado sold by date
nb_sold_by_date.plot(kind="line")
# Show the plot
plt.show()
```

```python
# Scatter plot of avg_price vs. nb_sold with title
avocado.plot(x="nb_sold",y="avg_price",kind="scatter",title="Number of avocado sold vs. average price")
# Show the plot

plt.show()
```

```python
# Histogram of conventional avg_price 
conv=avocado[avocado["type"]=="conventional"].hist(bins=10)
# Histogram of organic avg_price
organic=avocado[avocado["type"]=="organic"].hist(bins=20)
# Add a legend
plt.legend(["conv","organic"])
# Show the plot
plt.show()
```

```python
# Histogram of conventional avg_price 
avocado[avocado["type"] == "conventional"]["avg_price"].hist()

# Histogram of organic avg_price
avocado[avocado["type"] == "organic"]["avg_price"].hist()

# Add a legend
plt.legend(["conventional", "organic"])

# Show the plot
plt.show()
```

```python
# Modify bins to 20
avocado[avocado["type"] == "conventional"]["avg_price"].hist(alpha=0.5,bins=20)

# Modify bins to 20
avocado[avocado["type"] == "organic"]["avg_price"].hist(alpha=0.5,bins=20)

# Add a legend
plt.legend(["conventional", "organic"])

# Show the plot
plt.show()
```

```python
# Import matplotlib.pyplot with alias plt
import matplotlib.pyplot as plt

# Check individual values for missing values
print(avocado.isna())

# Check each column for missing values
print(avocado.isna().any())

# Bar plot of missing values by variable
avocado.isna().sum().plot(kind="bar")

# Show plot
plt.show()
```

```python
# Remove rows with missing values
avocados_complete = avocado.dropna()

# Check if any columns contain missing values
print(avocados_complete.isna().any())
```
