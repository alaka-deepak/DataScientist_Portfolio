2. Before writing any code, ask two questions:

Which column do we summarize? This depends on your business question. To count users, you look at user_id. To find revenue, you look at amount.
Which operation do we apply? This depends on the insight you need: sum() for totals, mean() or median() for typical values, min() and max() for extremes, count() or nunique() for counting.

3. Data aggregation is the data manipulation operation used to summarize data by applying operations to specific columns to produce single values.

4. Mean vs. Median

The mean is the sum of all values divided by the number of values.
The median is the middle value when data is sorted.
The median is not affected by outliers because it only considers the middle value(s), while the mean is influenced by outliers because it takes into account all values.
Choose median when you don't want outliers to affect your summary, and mean when there are no outliers or when you want to capture their effect.

```python
# NumPy is imported, seed is set

# Initialize random_walk
random_walk = [0]

# Complete the ___
for x in range(100) :
    # Set step: last element in random_walk
    step = random_walk[-1]

    # Roll the dice
    dice = np.random.randint(1,7)

    # Determine next step
    if dice <= 2:
        step = step - 1
    elif dice <= 5:
        step = step + 1
    else:
        step = step + np.random.randint(1,7)

    # append next_step to random_walk
    random_walk.append(step)

# Print random_walk
print(random_walk)

# Import matplotlib.pyplot as plt

import matplotlib.pyplot as plt


# Plot random_walk

plt.plot(random_walk)
# Show the plot
plt.show()
```

## LOC and ILOC

1. .loc → selects by label (row or column names).

2. .iloc → selects by integer position (0, 1, 2, …).

```python
# Add a year column to temperatures
temperatures['year']=temperatures["date"].dt.year

# Pivot avg_temp_c by country and city vs year
temp_by_country_city_vs_year = temperatures.pivot_table(
    values="avg_temp_c",
    index=['country','city'],
    columns="year"

)

# See the result
print(temp_by_country_city_vs_year)
```

```python
# Subset for Egypt to India
print(temp_by_country_city_vs_year.loc["Egypt":"India"])

# Subset for Egypt, Cairo to India, Delhi
print(temp_by_country_city_vs_year.loc[("Egypt","Cairo"):("India","Delhi")])

# Subset for Egypt, Cairo to India, Delhi, and 2005 to 2010
print(temp_by_country_city_vs_year.loc[("Egypt","Cairo"):("India","Delhi"),2022:2025])
```
