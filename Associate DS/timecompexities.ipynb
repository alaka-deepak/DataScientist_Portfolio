1. Big-O helps us understand how efficiently our code will scale.
2. Big-O notation tells you how the time or memory required by an algorithm grows as the input size n becomes larger.
3. 

| Big-O        | Meaning            | Typical example           |
| ------------ | ------------------ | ------------------------- |
| `O(1)`       | Constant           | Dictionary lookup         |
| `O(log n)`   | Logarithmic        | Binary search             |
| `O(n)`       | Linear             | Loop through a list       |
| `O(n log n)` | Linear-logarithmic | Efficient sorting         |
| `O(n²)`      | Quadratic          | Nested loops              |
| `O(2ⁿ)`      | Exponential        | Some recursive algorithms |


```python
## Examples O(1)

numbers = [10, 20, 30, 40, 50]

print(numbers[0])

# Complexity: O(1) — constant time
```

```python
# Example- O(n) - linear time( 5 times)

numbers = [10, 20, 30, 40, 50]

for number in numbers:
    print(number)
```

```python
## Example 3 — O(n²) -Quadratic time


for x in numbers:
    for y in numbers:
        print(x, y)
```
